```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros del problema
llegadas_por_minuto = 1 / 15  # 1 llegada cada 15 minutos
horas_operacion = 12
minutos_operacion = horas_operacion * 60

# Generar tiempos entre llegadas usando distribución exponencial
tiempo_entre_llegadas = np.random.exponential(scale=1/llegadas_por_minuto, size=1000)

# Filtrar valores dentro del rango de operación
tiempo_acumulado = np.cumsum(tiempo_entre_llegadas)
tiempo_acumulado = tiempo_acumulado[tiempo_acumulado <= minutos_operacion]

# Mostrar resultados
print(f"Número total de llegadas en {horas_operacion} horas: {len(tiempo_acumulado)}")
print("Tiempos entre llegadas:", tiempo_entre_llegadas[:10])

# Graficar histograma de tiempos entre llegadas
plt.hist(tiempo_entre_llegadas, bins=30, density=True, alpha=0.6, color='b', edgecolor='black')
plt.xlabel("Tiempo entre llegadas (minutos)")
plt.ylabel("Densidad")
plt.title("Distribución de tiempos entre llegadas de vehículos")
plt.grid(True)
plt.show()
```
