```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros del problema
media_espera = 10  # minutos
desviacion_espera = 2  # minutos
num_clientes = 50

# Generar tiempos de espera usando distribución lognormal
esperas = np.random.lognormal(mean=np.log(media_espera) - (desviacion_espera**2 / (2 * media_espera**2)),
                              sigma=desviacion_espera / media_espera,
                              size=num_clientes)

# Mostrar resultados
print(f"Tiempos de espera para {num_clientes} clientes:", esperas[:10])
print(f"Tiempo promedio simulado: {np.mean(esperas):.2f} minutos")
print(f"Desviación estándar simulada: {np.std(esperas):.2f} minutos")

# Graficar histograma de tiempos de espera
plt.hist(esperas, bins=15, density=True, alpha=0.6, color='r', edgecolor='black')
plt.xlabel("Tiempo de espera (minutos)")
plt.ylabel("Densidad")
plt.title("Distribución LogNormal de tiempos de espera en mostrador")
plt.grid(True)
plt.show()
```
